```python
# pip install pooled-ppi @ git+https://github.com/jurgjn/pooled-ppi@develop
import pooled_ppi.yeast_pools as yp
```

```python
# o13297 has one pair model, and one pool model
# filters argument (to select pairs) is passed on to pd.read_parquet
# foldcomp database identifiers are now {input_name}_{sample}_{chain_id}
# - input_name is either chain uniprot identifiers (for pairs) or hash of the chain identifiers (for pools)
# - sample is from AlphaFold3 (diffusion model sample)
# - chain_id is the original chain identifier in the AlphaFold3 mmcif; foldcomp does not support multimer entries so we're storing individual chains
af3_id = 'o13297'
yp.get_models(filters=[('af3_id1', '==', af3_id), ('af3_id2', '==', af3_id)])
```

```python
# extract first model (sample==0) from pair and pool
models = yp.get_models(filters=[('af3_id1', '==', af3_id), ('af3_id2', '==', af3_id), ('sample', '==', 0)])
models
```

```python
# use db_id1, db_id2 to save the individual heterodimers as PDB files
for i, row in models.iterrows():
    path = f'{row.af3_id1}_{row.af3_id2}_{row.input_type}_{row["sample"]}.pdb'
    yp.save_predictions_db([row.db_id1, row.db_id2], path)
    print(path, 'written')
```
