```python
# pip install pooled-ppi @ git+https://github.com/jurgjn/pooled-ppi@develop
from pprint import pprint
import pooled_ppi.yeast_pools as yp
```

```python
# Get 5 random models with the protein-of-interest (p0c2j7) from the table of all models
af3_id = 'p0c2j7'
pairs = yp.get_models(filters=[[('af3_id1', '==', af3_id)], [('af3_id2', '==', af3_id)]]).sample(5)
pairs
```

```python
# Get unique structure identifiers corresponding to af3_id
db_ids = set(pairs.query('af3_id1 == @af3_id')['db_id1'].tolist() + pairs.query('af3_id2 == @af3_id')['db_id2'].tolist())
pprint(sorted(db_ids))
```

```python
# use db_ids to load/save the files
for db_id in db_ids:
    path = f'{db_id}.pdb'
    yp.save_predictions_db([db_id], path)
    print(path, 'written')
```
